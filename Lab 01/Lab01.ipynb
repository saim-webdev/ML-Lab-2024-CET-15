```python
# task 08 //make funtion that give true if dog appear in given string
```

```python
def find_dog(text):
    return "dog" in text.lower()   # text is parameter that receive string
```

```python
print(find_dog("is there a dog here?"))
```

```python
print(find_dog("Dog is friendly animal"))
```

```python
print(find_dog("i have a cat"))
```

```python
# task 09 // make fution that count how many time dog appear in string
```

```python
def count_dog(text):
    return  text.lower().count("dog")
   
```

```python
# task 10 // create funtion khown as caught_speeding(speed, is_birthday)
```

```python
def caught_speeding(speed,is_birthday):
    if is_birthday:
        speed = speed-5
    if speed <= 60:
        return "no ticket"
    elif speed <= 80:
        return "small ticket"
    else:
        return "big ticket"
```

```python
print(caught_speeding(60,False))
```

```python
print(caught_speeding(81,False))
```

```python
print(caught_speeding(100,False))
```

```python
print(caught_speeding(40,False))
```
