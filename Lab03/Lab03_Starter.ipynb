## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python
# T3: your code here

# T3: Clean the Data

# T3(a) Missing Values
print("T3(a) - Missing Values Before Cleaning:")
print(df.isna().sum())

# Count rows before removing missing Brand or Model
rows_before = df.shape[0]

# Drop ads with missing Brand or Model
df = df.dropna(subset=["Brand", "Model"])

# Fill missing text values with Unknown
df["Registered City"] = df["Registered City"].fillna("Unknown")
df["Fuel"] = df["Fuel"].fillna("Unknown")
df["Transaction Type"] = df["Transaction Type"].fillna("Unknown")

# Print number of removed rows
rows_removed = rows_before - df.shape[0]
print("Rows Removed:", rows_removed)


# T3(b) Formatting
df["Registered City"] = df["Registered City"].str.strip().str.title()

df["Price"] = df["Price"].str.replace("Rs", "", regex=False)
df["Price"] = df["Price"].str.replace(",", "", regex=False)
df["Price"] = df["Price"].str.strip()

print("\nT3(b) - Formatting Completed:")
print(df[["Registered City", "Price"]].head())


# T3(c) Data Types
df["Price"] = df["Price"].astype("int64")
df["Year"] = df["Year"].astype("Int64")

print("\nT3(c) - Updated Data Types:")
print(df.dtypes)


# T3(d) Inconsistent Categories
cities_before = df["Registered City"].nunique()

df["Registered City"] = df["Registered City"].replace("Sukkar", "Sukkur")
df["Model"] = df["Model"].replace("Corrolla Altis", "Corolla Altis")

cities_after = df["Registered City"].nunique()

print("\nT3(d) - Inconsistent Categories Fixed:")
print("Different Cities Before:", cities_before)
print("Different Cities After:", cities_after)


# T3(e) Duplicates
duplicates = df.duplicated().sum()
print("\nT3(e) - Duplicate Ads:", duplicates)

df = df.drop_duplicates()

print("Rows After Removing Duplicates:", df.shape[0])


# T3(f) Invalid Values

# Set invalid KMs Driven values to NaN
df.loc[
    (df["KMs Driven"] < 100) |
    (df["KMs Driven"] > 1000000) |
    (df["KMs Driven"] == 12345) |
    (df["KMs Driven"] == 123456),
    "KMs Driven"
] = np.nan

# Identify fake prices and high-priced Suzuki/Daihatsu ads
invalid_prices = (
    (df["Price"] == 12345678) |
    (df["Price"] == 87654321) |
    (
        ((df["Brand"] == "Suzuki") | (df["Brand"] == "Daihatsu")) &
        (df["Price"] > 5000000)
    )
)

# Drop ads with invalid prices
df = df[~invalid_prices]

print("\nT3(f) - Invalid Values Fixed:")
print("Missing KMs Driven:", df["KMs Driven"].isna().sum())
print("Rows After Removing Invalid Price Ads:", df.shape[0])


# T3(g) Self-check
print("\nT3(g) - Final Dataset Shape:")
print(df.shape)

```

## Task 4 · Analyse
- **(a)** New column `Price_lakh` = price in lakh rupees (÷ 100,000).
- **(b)** New column `Age` = 2020 − `Year`.
- **(c)** The 5 most common brands.
- **(d)** The 5 cities with the most ads.
- **(e)** Median `Price_lakh` for each fuel type, highest first.
- **(f)** Median `Price_lakh` of a **Corolla GLI** for each year 2010–2018. How much value does it lose per year?
- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?
- **(h)** Number of ads and median price for Cash vs Installment/Leasing. Interpret in 1–2 lines.

```python
# T4: your code here
# T4: Data Analysis

# T4(a) Create Price_lakh column
df["Price_lakh"] = df["Price"] / 100000

print("T4(a) - Price in Lakh:")
print(df[["Price", "Price_lakh"]].head())


# T4(b) Create Age column
df["Age"] = 2020 - df["Year"]

print("\nT4(b) - Car Age:")
print(df[["Year", "Age"]].head())


# T4(c) Display 5 most common brands
print("\nT4(c) - Top 5 Most Common Brands:")
print(df["Brand"].value_counts().head(5))


# T4(d) Display 5 cities with the most advertisements
print("\nT4(d) - Top 5 Cities:")
print(df["Registered City"].value_counts().head(5))


# T4(e) Median Price_lakh for each fuel type
print("\nT4(e) - Median Price by Fuel Type:")
print(df.groupby("Fuel")["Price_lakh"].median()
      .sort_values(ascending=False))


# T4(f) Median price of Corolla GLI for each year (2010-2018)
corolla_gli = df[
    (df["Model"] == "Corolla GLI") &
    (df["Year"] >= 2010) &
    (df["Year"] <= 2018)
]

corolla_prices = corolla_gli.groupby("Year")["Price_lakh"].median()

print("\nT4(f) - Corolla GLI Median Prices:")
print(corolla_prices)

print("\nYearly Price Difference:")
print(corolla_prices.diff())


# T4(g) Display 5 most expensive advertisements
print("\nT4(g) - 5 Most Expensive Ads:")
print(df.nlargest(5, "Price")[["Brand", "Model", "Year", "Price"]])


# T4(h) Compare Cash and Installment/Leasing
print("\nT4(h) - Transaction Type Analysis:")
print(df.groupby("Transaction Type")["Price_lakh"]
      .agg(["count", "median"]))
```
