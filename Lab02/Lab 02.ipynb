```python

# Task 03: Randomness

# (a) Set seed to 2026 and generate one random value   it will make random values
np.random.seed(2026)
random_value = np.random.rand()
print("T3(a) One Random Value:", random_value)

# (b) Generate 25 random values from standard normal distribution
np.random.seed(2026)
random_numbers = np.random.randn(25)
print("T3(b) 25 Random Numbers:\n", random_numbers)

# (c) Roll a Ludo dice 10,000 times using seed 42
np.random.seed(42)
dice_rolls = np.random.randint(1, 7, 10000)
print("T3(c) First 20 Dice Rolls:", dice_rolls[:20])

# (d) Count how many times each dice face appears
faces, counts = np.unique(dice_rolls, return_counts=True)

print("T3(d) Dice Face Counts:")
print(np.column_stack((faces, counts)))

# (e) Estimate probability of getting a six and count double sixes
probability_six = np.mean(dice_rolls == 6)

# Double six means two consecutive rolls are both six
double_sixes = np.sum((dice_rolls[:-1] == 6) & (dice_rolls[1:] == 6))

print("T3(e) Estimated Probability of Six:", probability_six)
print("T3(e) Consecutive Double Sixes:", double_sixes)
```

```python
# Task 9: Grace Marks

# (a) Add 5 grace marks and cap at 100
grace_marks = np.clip(marks + 5, 0, 100)
print("T9(a) Marks After Grace:", grace_marks)

# (b) Calculate pass percentage
passed_students = grace_marks >= 50
pass_percentage = np.mean(passed_students) * 100
print("T9(b) Pass Percentage:", pass_percentage, "%")

# (c) Find students who passed only because of grace marks
grace_only_pass = np.where((marks < 50) & (grace_marks >= 50))[0]
print("T9(c) Students Passed Only Due to Grace Marks:", grace_only_pass)

# (d) Calculate z-scores
z_scores = (grace_marks - np.mean(grace_marks)) / np.std(grace_marks)
print("T9(d) Z-Scores:", np.round(z_scores, 2))
```
